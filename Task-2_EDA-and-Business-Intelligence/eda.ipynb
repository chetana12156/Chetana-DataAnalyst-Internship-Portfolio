Step 1 : Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2 : Load Dataset

```python
df = pd.read_csv('../data/cleaned_data.csv')  
df.head()
```

Step 3 : Dataset Overview

```python
df.shape
df.info()
df.describe()
```

Step 4 : Key Business Metrics (KPIs)

```python
total_revenue = df['Total_Spending'].sum()
avg_order_value = df['Total_Spending'].mean()
total_customers = len(df)

print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Average Spending: {avg_order_value:.2f}")
print(f"Total Customers: {total_customers}")
```

Step 5 : Top Products Analysis

```python

df.columns = df.columns.str.strip().str.lower()
product_spending = df[['mntwines','mntfruits','mntmeatproducts',
                       'mntfishproducts','mntsweetproducts','mntgoldprods']].sum()

product_spending.sort_values().plot(kind='barh')

import matplotlib.pyplot as plt
plt.title('Top Product Categories by Spending')
plt.xlabel('Total Spending')
plt.ylabel('Product Category')
plt.show()
```

Step 6 : Customer Segmentation

```python
sns.scatterplot(x='income', y='total_spending', data=df)
plt.title('Income vs Spending')
plt.show()
```

Step 7 : Recency Analysis

```python
df['recency'].plot(kind='hist', bins=20)

plt.title('Customer Recency Distribution')
plt.xlabel('Days Since Last Purchase')
plt.show()
```

Step 8 : Campaign Analysis

```python
df['response'].value_counts().plot(kind='bar')

plt.title('Campaign Response')
plt.show()
```

Step 10 : Correlation Analysis

```python
corr = df.corr(numeric_only=True)['total_spending'].sort_values()

corr.plot(kind='barh')

import matplotlib.pyplot as plt
plt.title('Correlation with Total Spending')
plt.show()
```

INSIGHTS 

###  Customer Behavior Insights

- A small group of customers contributes a large portion of total revenue, indicating the presence of high-value customers.
- Customers with higher income tend to spend more, showing a positive relationship between income and spending.
- Some high-income customers have low spending, representing an opportunity for targeted marketing.

---

###  Product Insights

- Wine and meat products generate the highest spending among all categories.
- Products like fruits and sweets have comparatively lower contribution.
- This indicates that customers prefer premium or high-value product categories.

---

###  Customer Activity Insights

- Customers with lower recency are more active and engaged.
- A significant number of customers have high recency, indicating inactivity.
- These inactive customers can be targeted with re-engagement campaigns.

---

###  Marketing Insights

- The majority of customers did not respond to marketing campaigns.
- This suggests that current marketing strategies may not be effective.
- There is a strong opportunity to improve targeting and personalization.

---

###  Overall Business Insight

- Revenue is driven by a combination of high-income customers and high-performing product categories.
- Customer segmentation and targeted marketing can significantly improve business performance.
