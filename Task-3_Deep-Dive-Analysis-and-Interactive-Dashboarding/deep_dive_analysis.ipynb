
## Customer Segmentation Analysis

Step 1 : Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2 : Load cleaned dataset

```python
df = pd.read_csv("C:\\Users\\cheta\\Desktop\\Customer_Analytics_&_Insights\\data\\cleaned_data.csv")

df.head()
```

Step 3 : Dataset Overview

```python
df.info()
df.describe()
```

Step 4 : KPI Calculations

```python
# KPI 1: Average Customer Spending

total_spending = df['Total_Spending'].sum()

total_customers = len(df)

avg_customer_spending = total_spending / total_customers

print("Average Customer Spending:", avg_customer_spending)

# KPI 2: Campaign Response Rate

responded_customers = df['Response'].sum()

response_rate = (responded_customers / total_customers) * 100

print("Campaign Response Rate:", response_rate, "%")

# KPI 3: Average Income

avg_income = df['Income'].mean()

print("Average Income:", avg_income)

# KPI 4: Web Purchase Rate

total_web = df['NumWebPurchases'].sum()

total_store = df['NumStorePurchases'].sum()

total_catalog = df['NumCatalogPurchases'].sum()

total_purchases = total_web + total_store + total_catalog

web_purchase_rate = (total_web / total_purchases) * 100

print("Web Purchase Rate:", web_purchase_rate, "%")

# KPI 5: Complaint Rate

total_complaints = df['Complain'].sum()

complaint_rate = (total_complaints / total_customers) * 100

print("Complaint Rate:", complaint_rate, "%")
```

Step 5 : Customer Segementation

```python
df['Customer_Segment'] = pd.qcut(
    df['Total_Spending'],
    q=3,
    labels=['Low Value', 'Medium Value', 'High Value']
)

df[['Customer_Segment', 'Total_Spending']].head()
```

Step 6 : Segment Calculations

```python
segment_analysis = df.groupby('Customer_Segment')[
    ['Income', 'Total_Spending']
].mean()

segment_analysis
```

Step 7 : Insights

 **KPI 1: Average Customer Spending**

Formula:

Average Customer Spending = Total Spending / Total Customers

This KPI measures the average revenue contribution per customer.

**KPI 2: Campaign Response Rate**

Formula:

Campaign Response Rate =
(Number of Customers Responded / Total Customers) × 100

This KPI measures the effectiveness of marketing campaigns.

**KPI 3: Average Income**

Formula:

Average Income =
Total Income / Total Customers

This KPI shows customer purchasing capability.

 **KPI 4: Web Purchase Rate**

Formula:

Web Purchase Rate =
(Total Web Purchases / Total Purchases) × 100

This KPI measures customer engagement through online channels.

**KPI 5: Complaint Rate**

Formula:

Complaint Rate =
(Number of Complaints / Total Customers) × 100

This KPI helps evaluate customer satisfaction.
